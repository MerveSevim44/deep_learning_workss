# Task 1: Scaler'ı yükleyiniz

```python
scaler = load("/content/sample_data/diabetes_scaler.joblib")
```

# Task 2: Modeli yükleyiniz

```python
model = load_model("/content/sample_data/diabetes_model.keras")
```

# Task 3: new_patients.csv dosyasını yükleyiniz

```python
new_patients = pd.read_csv("/content/sample_data/new_patients.csv")
new_patients.head()
```

# Task 4: Gerekli ön işleme işlemini yapınız

```python
X_scaled = scaler.transform(new_patients)
```

# Task 5: Model ile yeni hastaların diyabet durumlarını tahmin ediniz

```python
model.predict(X_scaled)
```
